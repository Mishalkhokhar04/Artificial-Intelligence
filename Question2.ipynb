```python

def count_vowels(text):
    count = 0
    for ch in text.lower():
        if ch == "a" or ch == "e" or ch == "i" or ch == "o" or ch == "u":
            count = count + 1
    return count


senten = input("Enter a sentence: ")
words = senten.split()        

total_words = len(words)
total_chars = len(senten)

lgt = words[0]
shtest = words[0]
for w in words:
    if len(w) > len(lgt):
        lgt = w
    if len(w) < len(shtest):
        shtest = w

vowels = count_vowels(senten)
uq_words = set(words)        
sort_words = sorted(uq_words)

print("Total words:", total_words)
print("Total characters:", total_chars)
print("Longest word:", lgt)
print("Shortest word:", shtest)
print("Vowels:", vowels)
print("Unique words:", uq_words)
print("Alphabetical order:", sort_words)

f = open("word_analysis.txt", "w")
f.write("Total words: " + str(total_words) + "\n")
f.write("Total characters: " + str(total_chars) + "\n")
f.write("Longest word: " + lgt + "\n")
f.write("Shortest word: " + shtest + "\n")
f.write("Vowels: " + str(vowels) + "\n")
f.write("Unique words: " + str(uq_words) + "\n")
f.write("Alphabetical order: " + str(sort_words) + "\n")
f.close()
print("Saved in word_analysis.txt")
```
