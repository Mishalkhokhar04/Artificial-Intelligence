```python


menu = {"Burger": 500, "Pizza": 800, "Fries": 250, "Drink": 150}
def get_price(price, quantity):
    return price * quantity


total = 0
bill_text = ""
more = "yes"

while more == "yes":
    print("Menu.")
    print("1. Burger Rs. 500")
    print("2. Pizza Rs. 800")
    print("3. Fries Rs. 250")
    print("4. Drink Rs. 150")

    choice = input("Select item (1-4): ")

    if choice == "1":
        item = "Burger."
    elif choice == "2":
        item = "Pizza."
    elif choice == "3":
        item = "Fries."
    elif choice == "4":
        item = "Drink."
    else:
        item = ""

    if item != "":
        quantity = int(input("Enter quantity: "))
        price = get_price(menu[item], quantity)
        total = total + price
        bill_text = bill_text + item + " x " + str(quantity) + " = Rs. " + str(price) + "\n"
    else:
        print("Wrong choice")

    more = input("Order more items? (yes/no): ")

bill_text = bill_text + "Final bill: Rs. " + str(total)

print("Your Bill:")
print(bill_text)

f = open("orders.txt", "w")
f.write(bill_text)
f.close()
print("Saved in orders.txt")
```
