```python
def get_result(python, ai, maths):
    total = python + ai + maths
    percentage = total / 300 * 100

    if percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"

    return total, percentage, grade


all_students = []  

for i in range(5):
    print("Student", i + 1)
    name = input("Enter name: ")
    roll = input("Enter roll number: ")
    python = int(input("Enter Python marks: "))
    ai = int(input("Enter AI marks: "))
    maths = int(input("Enter Maths marks: "))

    total, percentage, grade = get_result(python, ai, maths)

    student = {"name": name, "roll": roll, "python": python, "ai": ai,
               "maths": maths, "total": total,
               "percentage": percentage, "grade": grade}

    all_students.append(student)


print("Result Disply of all students:")
for s in all_students:
    print(s["name"], s["roll"], "Total:", s["total"],
          "Percentage:", s["percentage"], "Grade:", s["grade"])
top = all_students[0]
for s in all_students:
    if s["percentage"] > top["percentage"]:
        top = s

print("\nHighest percentage student:", top["name"], top["percentage"])

f = open("results.txt", "w")
for s in all_students:
    f.write(s["name"] + " " + s["roll"] + " Total: " + str(s["total"]) +
            " Percentage: " + str(s["percentage"]) + " Grade: " + s["grade"] + "\n")
f.write("Highest percentage student: " + top["name"] + " " + str(top["percentage"]) + "\n")
f.close()
print("Saved in results.txt")
```
