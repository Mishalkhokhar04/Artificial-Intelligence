```python

def calculate_bill(products):
    bill = 0
    for p in products:
        bill = bill + p["total"]
    return bill


products = []  

for i in range(5):
    print("Product", i + 1)
    name = input("Enter product name: ")
    price = float(input("Enter price: "))
    quantity = int(input("Enter quantity: "))

    product = {"name": name, "price": price, "quantity": quantity,
               "total": price * quantity}
    products.append(product)

bill = calculate_bill(products)

if bill > 10000:
    discount = bill * 0.10
else:
    discount = 0

final_bill = bill - discount


expensive = products[0]
for p in products:
    if p["price"] > expensive["price"]:
        expensive = p

print("Bill ")
for p in products:
    print(p["name"], p["price"], "x", p["quantity"], "=", p["total"])
print("Total bill:", bill)
print("Discount:", discount)
print("Final bill:", final_bill)
print("Most expensive product:", expensive["name"])

f = open("bill.txt", "w")
for p in products:
    f.write(p["name"] + " " + str(p["price"]) + " x " + str(p["quantity"]) +
            " = " + str(p["total"]) + "\n")
f.write("Total bill: " + str(bill) + "\n")
f.write("Discount: " + str(discount) + "\n")
f.write("Final bill: " + str(final_bill) + "\n")
f.write("Most expensive product: " + expensive["name"] + "\n")
f.close()
print("Saved in bill.txt")
```
